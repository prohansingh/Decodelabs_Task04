```python
# Remove duplicates
df = df.drop_duplicates()
```

```python
# Convert date column
df['Date'] = pd.to_datetime(df['Date'])

```

```python
# Create Month column
df['Month'] = df['Date'].dt.month_name()
```

```python
# Create Year column
df['Year'] = df['Date'].dt.year
```

```python
# Check missing values
print(df.isnull().sum())

```

```python
df.dropna(inplace=True)
```
